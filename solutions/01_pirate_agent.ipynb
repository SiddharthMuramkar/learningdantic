# Solution 01 — the pirate agent

Approach: the whole trick is the `instructions` string. You steer the persona in plain English.
We also print the usage: the dialect costs a few extra output tokens.

Notes:

- Run the same question twice and you'll get *different* answers, because models are not deterministic.
  Notebook 08 shows how to test without that (`FunctionModel`).
- If your answer had no dialect at all, make the `instructions` firmer. You'll meet this
  "harden the prompt" loop again in later exercises.

```python
from learningdantic import setup

MODEL = setup()
```

```python
from pydantic_ai import Agent

pirate_agent = Agent(
    MODEL,
    instructions=(
        'You answer every question in full pirate dialect: "Arrr", "matey", "be" instead '
        'of "is", and an "Arrr!" to close. Keep the actual answer accurate underneath the dialect.'
    ),
)

result = await pirate_agent.run("What is the difference between a list and a tuple in Python?")
print(result.output)
print()
print("dialect cost:", result.usage.output_tokens, "output tokens |", result.usage.cost, "USD")
```

```python
assert isinstance(result.output, str) and result.output.strip(), "the agent gave an empty answer"
assert "arr" in result.output.lower(), "no pirate dialect: make the instructions firmer"
assert result.usage.output_tokens > 0
print("all checks passed")
```
