# Solution 01 — the pirate agent

Approach: the *entire* trick is the `instructions` string — you're programming the persona.
Note we also print the usage: the dialect costs you a few extra output tokens.

```python
# ── SETUP — run this first in every notebook ─────────────────────────────────
import os
import sys

import logfire
from dotenv import find_dotenv, load_dotenv

# 1) Environment: walk up from notebooks/ to the repo root and load .env
load_dotenv(find_dotenv())

# 2) Observability: no Logfire account yet? This still works — data only ships
#    when a token exists (`uv run logfire auth`, or a LOGFIRE_TOKEN in .env).
logfire.configure(send_to_logfire="if-token-present")
logfire.instrument_pydantic_ai()  # v2: instrumentation is explicit — Agent(instrument=True) is gone

MODEL = os.getenv("MODEL", "openai:gpt-6.1-sol")

print("kernel :", sys.executable)  # should end in learningdantic/.venv/bin/python
print("MODEL  :", MODEL)
```

```python
from pydantic_ai import Agent

pirate_agent = Agent(
    MODEL,
    instructions=(
        'You answer every question in full pirate dialect: "Arrr", "matey", "be" instead '
        'of "is", and an "Arrr!" to close. Keep the actual answer accurate underneath the dialect.'
    ),
)

result = await pirate_agent.run("What is the difference between a list and a tuple in Python?")
print(result.output)
print()
print("dialect cost:", result.usage.output_tokens, "output tokens |", result.usage.cost)
```

**Notes**

- Run the same question twice and you'll get *different* answers — models are
  non-deterministic. Determinism is what notebook 08's `FunctionModel` is for.
- If your version didn't use dialect *at all*, the fix is a firmer `instructions`
  string — you'll meet this "prompt hardening" loop again in the exercises.
